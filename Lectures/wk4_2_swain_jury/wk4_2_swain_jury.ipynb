## Ways to simulate outcomes

Broadly, there are two types:

1) A limited number of outcomes

   - Simulate the outcomes
   - How?
      - Simulate one coin flip
      - There are two outcomes - heads/tails
      - Ask for a random number between X and Y
        - Two choices -- between 0 and 2 (so 0 or 1) -- each  must have an equal chance
    - Answer questions, e.g.
        - Flip 10 times.  How many heads in 10 coin flips?
        - How often are there 8/10 heads?

<br><p>

2) You are given the probability of an outcome
    - The percentage of black members of a community is 26%
    - Choose a person at random
    - Are they black or white?
    - How?
        - get a **decimal** number between 0 and 1
        - Is the number <= .26?
          - Person is black
        - Is the number > .26?
          - Person is white.

### The X boys problem: Simulate [num1] boys in families of size [num2]

```python
# Start with this cell
import numpy as np
import matplotlib.pyplot as plt
```

```python
num_of_boys = 2
family_size = 15
num_of_simulations = 10000
boys_in_families = np.zeros(num_of_simulations)

for i in np.arange(0,num_of_simulations):
    one_family = np.random.randint(0,2,family_size)
    one_num_of_boys = np.sum(one_family)
    boys_in_families[i] = one_num_of_boys
    # print(f"family: {one_family} num of boys: {one_num_of_boys}")

#print(boys_in_families)

#print(boys_in_families == num_of_boys)

num_X_boy_families = np.sum(boys_in_families == num_of_boys)
print(f"Number of {num_of_boys} boy families: {num_X_boy_families} (out of {num_of_simulations})")

prob = num_X_boy_families / num_of_simulations
print(f"Prob: {prob}")
```

```python
plt.hist(boys_in_families)
plt.axvline(2,color="red")
```

### The Swain Jury
- Will you simulate a fair jury or an unfair jury?
- How do you simulate one fair jury?
  - Remember the makeup of the community (26% vs 74%)
  - How do you simulate the community?
  - How do you pick 12 people at random from the community?
  - What is the difference between "with replacement" and "without replacement"?
  - Which do you want?
- One jury is only a single value.  What is the **range** of values you expect if the jury selection was fair?
- To simulate the fair world, what steps do you need to repeat?
- Which values will you store?
- How will you calculate the probability of a jury with 0 black members?

# steps in simulation

- 26% community was black. 
- 10-15% of panel members were black  
- 12 members. 

- simulate one jury member. 
  - 26% black 74% white. 
  - pick a number from 0-99 (1-100)  
    - 1-26 black person. 
    - 27-100 white person.

- repeat one jury member simulation 12 times
<br>

- repeat whole jury X times - X should be large 
  - simulate one jury
