  - simulation recipe
    - decide which world to simulate (usually the 'no difference' world)
    - set a number of repeats = X (make X small at first for checking and then big)
    - save space for one outcome for each repeat (the outcome array)
    - start a loop to run X times
      - simulate one outcome
      - calculate a critical value (e.g. number of boys)
      - save the outcome in your outcome array
    - count the number of outcomes
    - divide by the number of repeats
    - compare the 'no difference' world to the real world

### Can we put different coloured bars on the same plot?

```python
plt.hist(mac_fat,alpha=0.5,color="red")
plt.hist(sb_fat,alpha=0.5,color="blue")
plt.xlabel("Grams of fat")
plt.ylabel("Menu item counts")
```

### Why are red bars so much higher?

- There are just more items on the McDonalds menu
- Remember these are counts
- How can we put the two counts on the same scale?

### The 'density' option

`density = True` means plot the **proportion** of each sample that is in each bin

- This will put the heights on the same scale

```python
my_bins = np.arange(0,120,5)

plt.hist(mac_fat,alpha=0.5,color="red",bins=my_bins,density=True)
# alpha makes the bars transparent
# density plots the proportion of counts in each bin,
# so larger and smaller samples can be compared directly

plt.hist(sb_fat,alpha=0.5,color="blue",bins=my_bins,density=True)

plt.xlabel("grams of fat")
plt.ylabel("Count")
```
